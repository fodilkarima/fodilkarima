```python
import joblib

def prediction_billet_RL(data):

    # Copie des données pour ne pas modifier df
    resultats = data.copy()

    # On place l'identifiant du billet en index
    resultats = resultats.set_index("id")

    # Chargement du pipeline complet :
    # imputation + standardisation + RL
    pipeline_RL = joblib.load(
        "/content/drive/MyDrive/projet 12/modele_RL.joblib"
    )

    # Récupération des variables utilisées pendant l'entraînement
    variables = list(pipeline_RL.feature_names_in_)

    # Vérification des colonnes manquantes
    colonnes_manquantes = [
        colonne for colonne in variables
        if colonne not in resultats.columns
    ]

    if colonnes_manquantes:
        raise ValueError(
            f"Colonnes manquantes dans le fichier : {colonnes_manquantes}"
        )

    # Sélection des variables brutes dans le même ordre
    X = resultats[variables]

    # Le pipeline réalise automatiquement :
    # l'imputation, la standardisation et la prédiction
    prediction = pipeline_RL.predict(X)

    # Ajout des prédictions
    resultats["Prédiction"] = prediction

    # Conversion des résultats en texte
    resultats["Nature du billet"] = resultats["Prédiction"].map({
        1: "Authentique",
        0: "Contrefait"
    })

    return resultats
```

```python
#Charger le fichier des nouveaux billets
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/projet 12/test billets_production.csv')
df.head()
```
